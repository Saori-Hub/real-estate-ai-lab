
## Missing Value Strategy Evaluation

### 1. Missing Overview

* station_minutes: 1.53%
* building_age: 0.97%
* Slight concentration in Ogikubo station

The overall missing rate is low (<2%), suggesting limited structural data quality issues.

---

### 2. Imputation Strategy

A two-step imputation approach was applied:

1. Median imputation by nearest_station
2. Global median fallback

This prevents unnecessary data loss while preserving local station characteristics.

---

### 3. Model Comparison

Three approaches were compared:

* Drop missing rows
* Two-step imputation
* Two-step imputation + missing flags

Findings:

* Dropna achieved the highest R²
* Imputation reduced MAE
* Missing flags had negligible impact

---

### 4. Conclusion

Given the low missing rate, missingness itself does not appear to contain predictive information.

Two-step median imputation is sufficient for this dataset.

# 欠損値処理の検証（日本語）

### 1. 欠損状況

* station_minutes: 1.53%
* building_age: 0.97%
* 荻窪駅にやや偏りあり

欠損率は2%未満と低く、データ品質上の大きな問題は見られませんでした。

---

### 2. 補完戦略

以下の2段階補完を実施しました。

1. 駅別中央値による補完
2. 全体中央値によるフォールバック

これにより、駅特性を保ちながらデータ損失を回避しました。

---

### 3. モデル比較

* 欠損削除（Dropna）
* 2段階補完
* 2段階補完＋欠損フラグ

結果として、

* R²はDropnaが最も高い
* MAEは補完で改善
* 欠損フラグの影響はほぼなし

---

### 4. 結論

欠損率が低いため、欠損そのものは価格決定要因としての情報性は低いと考えられます。

本データセットでは、2段階中央値補完で十分と判断しました。
