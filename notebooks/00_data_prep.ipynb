```python
import os

for root, dirs, files in os.walk(".", topdown=True):
    level = root.count(os.sep)
    indent = " " * 4 * level
    print(f"{indent}{os.path.basename(root)}/")
    subindent = " " * 4 * (level + 1)
    for f in files:
        print(f"{subindent}{f}")
```

```python
import os

for root, dirs, files in os.walk("data", topdown=True):
    level = root.count(os.sep)
    indent = " " * 4 * level
    print(f"{indent}{root}/")
    for f in files:
        print(f"{indent}    {f}")
```
